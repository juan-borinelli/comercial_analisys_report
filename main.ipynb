```python
import pandas as pd
```

```python
df_site = pd.read_csv("base_midianet_site.csv")
df_radio = pd.read_csv("base_radio_conecta.csv")
df_tv = pd.read_csv("base_tv_panorama.csv")
```

# ANALISE ISOLADA SITE

```python
df_site
```

```python
med_site = df_site['valor_ultima_campanha'].mean()
med_camp_site = df_site['qtd_campanhas'].mean()
```

```python
# CLIENTES SITE COM MAIS CAMPANHAS DO QUE A MÉDIA
c_site_recorrente = df_site[df_site['qtd_campanhas'] > med_camp_site].sort_values(by='qtd_campanhas', ascending=False)
c_site_recorrente
```

```python
c_site_valor_alto = df_site[df_site['valor_ultima_campanha'] > med_site].sort_values(by='valor_ultima_campanha', ascending=False)
c_site_valor_alto
```

```python
c_site_ouro = pd.merge(c_site_recorrente, c_site_valor_alto, how='inner')
c_site_ouro
```

```python
# CLIENTES OURO INATIVOS (FAZER PROPOSTA DE REATIVAÇÃO)
c_site_ouro[c_site_ouro['status'] == 'Inativo']
```
